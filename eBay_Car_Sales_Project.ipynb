**3.** **Columns that should be converted to numeric dtypes are** 'price' and 'odometer'. These columns contain numeric values that are currently string dtypes but should be integer dtypes. We will first remove any non-numeric characters and then convert the data types to integers so future calculations are possible.

```python
#convert string dtype to numeric
autos['price'] = autos['price'].astype(int)

autos['odometer'] = autos['odometer'].astype(int)

print(f'price dtype: {autos['price'].dtype}')
print(f'odometer dtype: {autos['odometer'].dtype}')
```

```python
#rename 'odometer' column
autos.rename({'odometer': 'odometer_km'}, axis=1, inplace=True)

od_describe = autos['odometer_km'].describe()
od_describe
```

### Incorrect Registration Year Data

As we established, some of the date values in our 'registration_year' column are inaccurate. We automatically know that values can't exceed the year 2016 since a car can't be registed before the car was seen. To start our analysis we will look at the number of values outside of the range 1900-2016 and see what returns. 

```python
#create a copy of the dataset for analysis
autos_filt_copy = autos_filtered.copy()

#look at the values above 2016 and below 1900
outliers = autos_filt_copy[(autos_filt_copy['registration_year'] < 1900) | (autos_filt_copy['registration_year'] > 2016)]

outliers['registration_year']
```

```python
#number of times each year appears
outliers['registration_year'].value_counts()
```

It looks like most of the incorrect dates are registered for 2017 and 2018. Nonetheless, these values cannot reasonable be correct since the dataset was collected in 2016. Our next task will be to remove values below 1900 and above 2016. 

```python
#use the origional dataset to keep only the entries between 1900 and 2016
autos_filt = autos_filtered[(autos_filtered['registration_year'].between(1900, 2016))]

#count the number of times each value is in the dataset and sort based on year from closer to farther
sort_ry = autos_filt['registration_year'].value_counts(normalize=True).sort_index(ascending=False)

sort_ry
```

```python
#descriptive statistics on sort_ry
sort_ry.describe()
```

We now have a much smaller standard deviation of about 2%.

## Dataset Analysis

The final part this this project is a small exploration of the mean price of top brands sold on eBay.

### Exploring Price by Brand

As stated above our analysis will consist of exploring the mean price of top car brands sold on eBay. To start we will count the number of times each brand appears in the dataset and then find the mean price of the top brands. 

```python
#count the number of times each brand appears (value_counts automatically sorts highest to lowest)
brand_sort = autos_filt['brand'].value_counts(normalize=True)

brand_sort.head(20)
```

We will use the top 4% of brands for our analysis since the percent makeup of brands under the 4% marker are closer in distribution. That leaves 7 vehicles for our analysis: Volkswagen, BMW, Opel, Mercedes Benz, Auid, Ford, and Renault.

Our next task is to seperate the top 4% for our analysis. 

```python
#from the brand column select only those that make up 4% or more of the sales by brand
top_4_perc = brand_sort[brand_sort >= 0.04].index

top_4_perc
```

Finally, we will calculate the mean price of the top 4% and order based on cost (most expensive at the top).

```python
#create a dictionary that holds the car brand and mean price
mean_brand_price = {}

for brand in top_4_perc:
    key = autos_filt[autos_filt['brand'] == brand]
    mean_brand_price[brand] = int(key['price'].mean())

#sort dictionary by converting key-value-pairs to tuples and order given index 1 of tuple - descending
dic_sorted = sorted(mean_brand_price.items(), key=lambda x: x[1], reverse=True)

dic_sorted
```

On average the most expensive vehicles are from Audi and the cheapest are from Renault. The top 4% of cars sold on eBay have a wide price range. 

### Storing Aggregate Data in a DataFrame

Our last task will be to store our mean price by brand data in a DataFrame so we can look at other aggregate data side-by-side. The process is simply to convert the dictionary into a Series using the `.Series()` attribute and then making it a two-dimensional DataFrame using the `.DataFrame()` attribute.

```python
#convert dictionary to pandas Series
brand_price_series = pd.Series(mean_brand_price)
brand_price_series
```

```python
#create a DataFrame from pandas Series
brand_price_df = pd.DataFrame(brand_price_series, columns=['mean_price'])
brand_price_df
```

To demonstrate the benefit of holding aggregate data in a DataFrame, we will quickly calculate the mean mileage of the top 4% of vehicle by brand. We will then turn the dictionary into a Series and add it to out 'brand_price_df' DataFrame so we can view the aggregated data side-by-side.

```python
#create a dictionary that holds the car brand and mean mileage
mean_brand_mileage = {}

for brand in top_4_perc:
    key = autos_filt[autos_filt['brand'] == brand]
    mean_brand_mileage[brand] = int(key['odometer_km'].mean())

#sort dictionary by converting key-value-pairs to tuples and order given index 1 of tuple - descending
dic = sorted(mean_brand_mileage.items(), key=lambda x: x[1], reverse=True)

dic    
```

```python
#convert dictionary to pandas Series
brand_mileage = pd.Series(mean_brand_mileage)
brand_mileage
```

```python
#assign the new 'brand_mileage' Series to our existing 'brand_price_df' DataFrame
brand_price_df['brand_mileage'] = brand_mileage

#rename 'brand_mileage' to 'mean_mileage'
brand_price_df.rename({'brand_mileage': 'mean_mileage'}, axis=1, inplace=True)
brand_price_df
```

Now we can see our 'mean_price' and 'mean_mileage' by brand side-by-side. We could use this to preform basic statistical analysis between columns or within columns. Given the output, it appears that most cars, independent of brand and price, have a similar mean mileage between 120,000km and 135,000km.
